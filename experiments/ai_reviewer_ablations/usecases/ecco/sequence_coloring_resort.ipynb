# ECCO p02973 Sequence coloring
#### Payload type: CPU-bound Python loop re-sorting a 10^5-element list on every input element

Source: [ECCO edit/test row 705](https://huggingface.co/datasets/CodeEff/ECCO/viewer/edit/test?row=705) - the slow submission, verbatim, wrapped as `solve(text)`.

```python
import random
import zlib
```

```python
# Inputs scaled within the problem's constraints, fixed by seed
def generate_input(seed, length=30000):
    """N values in [0, 10^9]; the constraints allow N up to 10^5."""
    rng = random.Random(seed)
    values = [str(rng.randint(0, 10**9)) for _ in range(length)]
    return f"{length}\n" + "\n".join(values) + "\n"


scaled_inputs = [generate_input(seed) for seed in range(4)]
```

```python
import bisect


def solve(text):
    lines = iter(text.splitlines())
    input = lambda: next(lines)

    N=int(eval(input()))

    line=[-1]*10**5

    for _ in range(N):
        A=int(eval(input()))
        index=bisect.bisect_left(line,A)
        if index==0:
            line[0]=A
        else:
            line[index-1]=A
        line.sort()
        #print(line)
    ans=[]
    while line:
        num=line.pop(-1)
        if num==-1:
            break
        else:
            ans.append(num)

    return str(len(ans))


answers = [solve(text) for text in scaled_inputs]
checksum = sum(zlib.crc32(answer.encode()) for answer in answers)
```
